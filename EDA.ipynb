```python
sns.countplot(data=df,x="species")
plt.title("number of flowers by species")
plt.xlabel("species")
plt.ylabel("count")
plt.show()
```

```python
sns.histplot(
    data=df,x="sepal_length",
    bins=10,
    kde=True
)
```

```python
sns.boxplot(
    data=df,
    x="species",
    y="petal_length"
)
plt.title("petal length by species")
plt.show()
```
